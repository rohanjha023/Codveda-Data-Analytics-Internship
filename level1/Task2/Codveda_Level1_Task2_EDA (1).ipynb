## **Level 1  Task 2: Exploratory Data Analysis (EDA)**

**1. Import Libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

**2. View Data (Head)**

```python
df = pd.read_csv("churn_cleaned.csv")
df.head()
```

```python
print("Shape of dataset:", df.shape)
print("\nColumns:\n", df.columns)
print("\nData Types:\n", df.dtypes)
print("\nMissing Values:\n", df.isnull().sum())
```

**The cleaned dataset contains 2666 rows and 20 columns. It has no major missing value issues after preprocessing and is ready for exploratory data analysis.**

**Summary Statistics**

```python
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols].describe()
```

```python
df[numeric_cols].mean()
```

```python
df[numeric_cols].median()
```

```python
df[numeric_cols].mode().iloc[0]
```

```python
df[numeric_cols].std()
```

**Categorical Analysis**

```python
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    print(f"\nValue counts for {col}:")
    print(df[col].value_counts())
```

**Churn Distribution**

```python
print(df['churn'].value_counts())
print(df['churn'].value_counts(normalize=True) * 100)
```

**Most customers are non-churn customers, which indicates the dataset is slightly imbalanced toward retained customers.**

Histograms

```python
important_cols = [
    'account_length',
    'total_day_minutes',
    'total_eve_minutes',
    'total_night_minutes',
    'total_intl_minutes',
    'customer_service_calls'
]

for col in important_cols:
    plt.figure(figsize=(8,4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

**Boxplots**

```python
for col in important_cols:
    plt.figure(figsize=(8,4))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

**Histograms help understand the distribution of numeric features, while boxplots help detect outliers. Features like call minutes and customer service calls may contain some extreme observations.**

**Churn-based Analysis**

```python
plt.figure(figsize=(6,4))
sns.countplot(x='churn', data=df)
plt.title("Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='international_plan', hue='churn', data=df)
plt.title("International Plan vs Churn")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='voice_mail_plan', hue='churn', data=df)
plt.title("Voice Mail Plan vs Churn")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.countplot(x='customer_service_calls', hue='churn', data=df)
plt.title("Customer Service Calls vs Churn")
plt.show()
```

**Customers with higher customer service calls seem more likely to churn.
Customers having an international plan may show a higher churn tendency compared to those without it.**

**Scatter Plots**

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x='total_day_minutes', y='total_day_charge', hue='churn', data=df)
plt.title("Total Day Minutes vs Total Day Charge")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x='total_eve_minutes', y='total_eve_charge', hue='churn', data=df)
plt.title("Total Evening Minutes vs Total Evening Charge")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x='total_night_minutes', y='total_night_charge', hue='churn', data=df)
plt.title("Total Night Minutes vs Total Night Charge")
plt.show()
```

**Charges and minutes are directly proportional, so these pairs are expected to have very high positive correlation.**

**Correlation Heatmap**

```python
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(14,8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap")
plt.show()
```

**Grouped Analysis**

```python
df.groupby('churn')[numeric_cols].mean()
```

```python
df.groupby('churn')[[
    'total_day_minutes',
    'total_eve_minutes',
    'total_night_minutes',
    'total_intl_minutes',
    'customer_service_calls'
]].mean()
```

**Final Insights**

1.  The dataset contains 2666 cleaned customer records and is suitable for analysis

2.  Most customers did not churn, showing an imbalanced target distribution.

3. Usage-related variables such as total day, evening, and night minutes are strongly correlated with their corresponding charges.

4. Customers with higher customer service calls appear more likely to churn.

5. International plan customers may have a relatively higher churn tendency.

6. Boxplots indicate the presence of some outliers in usage and service-related features.

7. Correlation analysis shows very strong linear relationships between minutes and charges.
